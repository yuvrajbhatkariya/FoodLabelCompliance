```python
import os
from dotenv import load_dotenv
from google import genai

# Load the API key from your .env file
load_dotenv()
api_key = os.environ.get("GEMINI_API_KEY")
model_name = os.environ.get("GEMINI_MODEL", "gemini-3.8-flash")

if not api_key:
    print("❌ Error: GEMINI_API_KEY not found in .env file.")
    exit(1)

print(f"Testing connection to {model_name}...")

try:
    # Initialize the client
    client = genai.Client(api_key=api_key)
    
    # Send a simple text prompt
    response = client.models.generate_content(
        model=model_name, 
        contents="Hello! This is a test. Reply with 'API is working' if you receive this."
    )
    
    print("\n✅ SUCCESS! The API is working.")
    print(f"Model replied: {response.text}")

except Exception as e:
    print("\n❌ FAILED! There was an error connecting to the API.")
    print(f"Error details: {e}")
```

```python
import os
from dotenv import load_dotenv
from google import genai
from google.genai import types

load_dotenv(override=True)
api_key = os.environ.get("GEMINI_API_KEY")
model_name = os.environ.get("GEMINI_MODEL")

image_path = "test_images/t2.jpeg"

if not os.path.exists(image_path):
    print(f"❌ Error: Please place an image named '{image_path}' in this folder.")
    exit(1)

print(f"Testing Vision API with {model_name}...")

try:
    client = genai.Client(api_key=api_key)
    
    # 1. Read the image as raw bytes (exactly how FastAPI receives uploads)
    with open(image_path, "rb") as f:
        image_bytes = f.read()
        
    # 2. Package it for the Gemini API
    image_part = types.Part.from_bytes(
        data=image_bytes, 
        mime_type="image/jpeg"
    )
    
    # 3. Send both the image and a text prompt
    response = client.models.generate_content(
        model=model_name,
        contents=[
            image_part,
            "What product is in this image? Reply with just the brand and product name."
        ]
    )
    
    print("\n✅ SUCCESS! The Vision API is working.")
    print(f"Model replied: {response.text}")

except Exception as e:
    print("\n❌ FAILED! There was an error processing the image.")
    print(f"Error details: {e}")
```
